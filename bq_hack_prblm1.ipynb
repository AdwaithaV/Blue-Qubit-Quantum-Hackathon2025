```python
from qiskit import QuantumCircuit

qc = QuantumCircuit.from_qasm_file("/content/P2_swift_rise.qasm")
print(qc.draw())
```

```python
import bluequbit

qiskit_circuit = qc
bq = bluequbit.init("qxQbgjeVNI7Q2xLi2ZDsaYtnz5ldVRSv")
result = bq.run(qiskit_circuit, device='gpu',shots=2000)#increased the number of shots as the number of qubits were higher
```

```python
counts = result.get_counts()
print(counts)
```

```python
# Get all results sorted by counts (descending)
sorted_counts = sorted(counts.items(), key=lambda x: x[1], reverse=True)

# Get the maximum
max_bitstring, max_counts = sorted_counts[0]
print(f"Top result: {max_bitstring} with {max_counts} counts")

# Get top 3 results
for i, (bitstring, count) in enumerate(sorted_counts[:3]):
    print(f"Rank {i+1}: {bitstring} ({count} counts)")
```

```python
from qiskit import QuantumCircuit
from qiskit.visualization import circuit_drawer
qc = QuantumCircuit.from_qasm_file("/content/P3__sharp_peak.qasm")
qc.draw(output='mpl')
```

```python
import bluequbit

qiskit_circuit = qc
bq = bluequbit.init("qxQbgjeVNI7Q2xLi2ZDsaYtnz5ldVRSv")
result = bq.run(qiskit_circuit, device='mps.cpu',shots=10000)
```

```python
counts = result.get_counts()
print(counts)
```

```python
# Get all results sorted by counts (descending)
sorted_counts = sorted(counts.items(), key=lambda x: x[1], reverse=True)

# Get the maximum
max_bitstring, max_counts = sorted_counts[0]
print(f"Top result: {max_bitstring} with {max_counts} counts")

# Get top 3 results
for i, (bitstring, count) in enumerate(sorted_counts[:3]):
    print(f"Rank {i+1}: {bitstring} ({count} counts)")
```
